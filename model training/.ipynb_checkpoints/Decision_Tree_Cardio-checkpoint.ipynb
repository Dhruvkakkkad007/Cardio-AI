# import necessary libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder

import warnings
warnings.filterwarnings('ignore')
```

# Import cardio_cleaned.csv data set

```python
df = pd.read_csv('cardio_cleaned.csv')
df.head()
```

```python
df.info()
```

# Check the distribution of the target

```python
print(df['cardio'].value_counts())
```

# Check for missing values

```python
df.isnull().sum()
```

# Visualize Distributions

```python
df.hist(figsize=(14, 12))
plt.tight_layout()
plt.show()
```

# Convert Target data into integer code

```python
df['cardio'].unique()
```

```python
le = LabelEncoder()
df['cardio'] = le.fit_transform(df['cardio'])
df['cardio']
```

# Divide the data into input and output

```python
# Axis = 1 means drop column, axis = 0 means drop row
# Drop 'cardio' (target) and 'id' (identifier) from features
if 'id' in df.columns:
    x = df.drop(columns=['cardio', 'id'])
else:
    x = df.drop('cardio', axis=1)

y = df['cardio']

# shape gives the number of rows and columns in the dataset
print("Input features shape (X):", x.shape)
print("Target shape (y):", y.shape)
```

# Splitting the dataset into the Training set and Test set

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

# Create Model

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(random_state=42)
```

# Fitting DecisionTreeClassifier on dataset across different depths

```python
train_accs = []
test_accs = []

for depth in range(1, 11):
    temp_model = DecisionTreeClassifier(max_depth=depth, random_state=42)
    temp_model.fit(X_train, y_train)
    train_acc = temp_model.score(X_train, y_train)
    test_acc = temp_model.score(X_test, y_test)
    train_accs.append(train_acc)
    test_accs.append(test_acc)
    print(f"Depth: {depth:2d} | Training Accuracy: {train_acc:.4f} | Testing Accuracy: {test_acc:.4f}")
```

# Train Final Decision Tree Model

```python
# Training with optimal depth for balance between performance and interpretability
model = DecisionTreeClassifier(max_depth=4, random_state=42)
model.fit(X_train, y_train)
```

# Display Decision Tree

```python
plt.figure(figsize=(24, 12), dpi=300)
plot_tree(
    model,
    filled=True,
    feature_names=list(x.columns),
    class_names=['No Cardio Disease', 'Cardio Disease'],
    rounded=True,
    fontsize=10
)
plt.title("Decision Tree Visualization (Max Depth = 4)", fontsize=16)
plt.tight_layout()
plt.show()
```

# Predict the x_test

```python
y_pred = model.predict(X_test)
print("Predictions on Test Set:")
print(y_pred)
```

# Display Training Accuracy

```python
TrainingAccuracy = model.score(X_train, y_train)
print(f"Training Accuracy: {TrainingAccuracy:.4f}")
TrainingAccuracy
```

# Display Test Accuracy

```python
testAccuracy = model.score(X_test, y_test)
print(f"Testing Accuracy: {testAccuracy:.4f}")
testAccuracy
```

# Classification Report & Performance Metrics

```python
print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=['No Cardio Disease', 'Cardio Disease']))

print("\nConfusion Matrix:")
cm = confusion_matrix(y_test, y_pred)
print(cm)
```
